```python
message = "Привет! это послание из прошлого, передай его!"
key = 1000
print(message)

secured_message = ""

for i in message:
    secured_message += chr(ord(i)+key)

print(secured_message)

print("Дешивровка....")
for i in secured_message:
    print(chr(ord(i)-key), end="")
```
