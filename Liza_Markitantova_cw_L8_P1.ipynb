```python
for i in range(5):
    try:
        # ZeroDivisionError ValueError
        cash = int(input("Сумму которую хочешь накопить"))
        n = int(input("Кол-во месяцев за которое хочешь накопить:"))
        res = cash /n
        print(f"Тебе нужно откладывать {res} в месяц, чтобы за месяцев {n} накопить {cash}")
    except ZeroDivisionError as e:
        print("Делить на ноль нельзя")
        print(e)
        print(e.args)
        print(e.with_traceback)
    except ValueError as e:
        print("Не пытайся конвертировать строку из букв в int")
        print(e)
        print(e.args)
        print(e.with_traceback)
    except:
        print("Возникла ошибка, попробуй еще раз")
```
