```python
def GiveMeTheMessage(text):
    message = input(text)

    return message


def zashifrovatb(message, k):
    print("Шифровка....")
    secured_message = ""

    for i in message:
        secured_message += chr(ord(i)+k)

    return secured_message


def deshifrovka(secured_message, k):
    print("Дешивровка....")
    open_message = ""

    for i in secured_message:
        open_message += chr(ord(i)-k)

    return open_message


def runApp():
    menu = """
_________________________________________________________
|                 Шифровалка v1.0
|      1 - зашифровать            @@@@@#####
|      2 - дешифровать            %%&%&%&&%*%(%&
|      0 - выход                  *#$(&%$(*^@*(^%(
|________________________________________________________

----->"""

    userInput = int(input(menu))

    while userInput != 0:
        key = int(input("Введи ключ для шифра(число целое не отрицательное):"))
        msg = GiveMeTheMessage("Введи сообщение для послания которе нужно обработать:")

        if userInput == 1:
            res_msg = zashifrovatb(msg, key)
            print(res_msg)
        elif userInput == 2:
                de_res_msg = deshifrovka(res_msg, key)
                print(de_res_msg)
        else:
            print("Я не понял тебя. повтори ввод заново")

        userInput = int(input("1 - зашифровать, 2 - дешифровать, 0 - выход"))


runApp()
```

```python
# Lab 5.3 --> factorial

# n = 5 фактриал = 1*2*3*4*5
# n = 2 фактриал = 1*2
# n = 9 фактриал = 1*2*3*4*5*6*7*8*9
# n = 1 фактриал = 1
# n = 0 фактриал = 1
# n = -999 фактриал = False

def myFactorial(n):
    if n < 0:
        return False
    if n <= 1:
        return 1

    target_value = 1
    for i in range(2, n+1):
        target_value *= i
        
    return target_value


def run(): # !!!!
    li = [-999, 0, 1, 9, 2, 5]
    resList = [False, 1, 1, 362880, 2, 121]
    
    for number, result in zip(li, resList):
        if myFactorial(number) == result:
            print("Ok")
        else:
            print("failed")
            print("Ожидалость:", number, result)
            print("А получилось:", number, myFactorial(number))


run() # !!!!
```
