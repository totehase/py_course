```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"user: {self.name} {self.age}"
    def __repr__(self):
        return f"user: {self.name} {self.age}"
```

```python
d1 = User("D", 1)
```

```python
k1 = User("K", 1)
```
