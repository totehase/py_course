```python
hp = 100
time_limit = 60
while time_limit >= 0:
    print("time_limit", time_limit)
    print("у вас осталось hp: ", hp)
    
    if time_limit % 2 ==0:
    # получили урон в голову
        hp -= 10
        print("Вы получили удар в голову -10 hp")
    if time_limit % 2 ==0:
    # получили урон в ногу
        hp -= 5
        print("Вы получили удар в голову -5 hp")
    

    print()
    time_limit -=1
```
