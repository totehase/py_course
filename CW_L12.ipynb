```python
class pow_two:
    def __init__(self, maxi=0):
        self.maxi = maxi

    def __iter__(self):
        self.counter = 0

        return self
    
    def __next__(self):
        if self.counter <= self.maxi:
            result = 2 ** self.counter
            self.counter += 1

            return result

    raise StopIteration
```

```python
class Tasks(list):
    def my_append(self, value):
        print("добавляю в список значение", value)
        self.append(value)
    def my_pop(self):
        if len(self) == 0:
            print("Список пуст")
            return
        res = self.pop()
        print("удаляю значение из списка", res)
```

```python
ttt = Tasks()
```

```python
ttt.my_append(55)
```

```python
ttt
```

```python
ttt.my_append(5)
```

```python
ttt
```

```python
My_tasks = type(
    "My_tasks",
    (list,),
    dict(
        my_append=lambda self, value: self.append(value),
        my_pop=lambda self: self.pop(),
    )
)
```

```python
my_ttt = My_tasks()
```
